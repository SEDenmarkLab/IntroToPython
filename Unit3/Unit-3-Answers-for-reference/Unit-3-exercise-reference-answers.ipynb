```python
def median(lists):
    sortedList = sorted(lists)
    n = len(sortedList)
    mid = n // 2
    if n % 2 != 0:
        return sortedList[mid]
    else:
        return (sortedList[mid - 1] + sortedList[mid]) / 2
    
print(median([10.5, 2, 33, 4, 15]))
```

```python
def bubbleSort(lists):

    swapCount = 0
    
    if len(lists) <= 1:
        return swapCount
    
    while True:
        swapped = False
        for i in range(len(lists) - 1):
            if lists[i] > lists[i + 1]:
                lists[i], lists[i + 1] = lists[i + 1], lists[i]
                swapCount += 1
                swapped = True
        if not swapped:
            break
    return swapCount

lists = [10.5, 2, 33, 4, 15]

print(bubbleSort(lists))
print(lists)
```

```python
import time

def recursiveLoop(currentDepth, currentSize):
        if currentDepth == 0:
            time.sleep(0.00001) 
            return
        for i in range(currentSize):
            recursiveLoop(currentDepth - 1, currentSize)

def nestedLoopTime(depth, size):

    wallTimeStart = time.perf_counter()
    cpuTimeStart = time.process_time()

    recursiveLoop(depth, size)

    wallTimeEnd = time.perf_counter()
    cpuTimeEnd = time.process_time()

    print("wall time : " + str(wallTimeEnd - wallTimeStart))
    print("cpu time : " + str(cpuTimeEnd - cpuTimeStart))

nestedLoopTime(5, 10)
```
