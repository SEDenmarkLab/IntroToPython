```python
from typing import List
def sortBySecondLetter(names: List[str]) -> List[str]:
    """sorting a list of name by second letter

    Parameters
    ----------
    names : List[str]
        Unsorted list of names

    Returns
    -------
    List[str]
        Sorted list of names
    """
    if not isinstance(names, list):
        raise TypeError("input needs to be a list")
    if len(names) < 2:
        raise ValueError("there is no reason to sort a list with less than two element")
    for i in names:
        if not isinstance(i, str):
            raise TypeError("elements in list needs to be string")
        if len(i) < 2:
            raise ValueError("each name need to be more than two character in order to sort it by the second letter")

    return sorted(names, key=lambda x : x[1])

print(sortBySecondLetter(['Matthew', 'Tatsuto', 'Dongchen', 'Tony', 'Blake', 'Elena', 'Shangheng', 'Margherita', 'Matt', 'Jim', 'Nick', 'Jacob', 'Roberto', 'Felix', 'Michela', 'Yuki', 'Behrad', 'Sateesh']))
```

```python
from typing import List

def convertUnitsToSi(distances: List[str]) -> List[str]:
    """coverting unit to SI with scientific notation

    Parameters
    ----------
    distances : List[str]
        list of distance strings ready to be converted, expected in the format "Value Unit" (e.g., "134 nm").

    Returns
    -------
    List[str]
        list of converted string
    """
    if not isinstance(distances, list):
        raise TypeError("input shoud be a list")
    conversonDict = {'A': 1e-10, 'nm': 1e-9, 'um': 1e-6, 'mm': 1e-3, 'cm': 1e-2, 'm': 1e0, 'km': 1e3}
    result = []
    for i in distances:
        if not isinstance(i, str):
            raise TypeError("elements in list should be string")
        parts = i.split()
        if not len(parts) == 2:
            raise ValueError("elements in list must be in the form of 'Value Unit'")
        value = float(parts[0])
        unit = parts[1]
        if unit not in conversonDict:
            raise ValueError("unit not supported")
        meter = value * conversonDict[unit]
        result.append(f"{meter:.2e} m")
    return result

print(convertUnitsToSi(['1.4 A', '134 nm', '254 um', '530 mm', '765 cm', '812 m', '299 km']))
```

```python
from typing import Union

def sumThatSupportInfiniteInputs(*nums: Union[int, float, complex]) -> Union[int, float, complex]:
    """a sum function that handles an undefined number of inputs including int, float, and complex numbers 


    Parameters
    ----------
    *nums : Union[int, float, complex]
        numbers to be summed.

    Returns
    -------
    Union[int, float, complex]
        Returning the sum of the input numbers
    """
    
    if len(nums) == 0:
        raise ValueError("empty inputs")
    if len(nums) == 1:
        raise ValueError("there is no reason to sum a single number")
    result = 0
    for i in nums:
        if not isinstance(i, (int, float, complex)):
            raise TypeError("each input number should either be int, float, or complex")
        result += i
    return result

print(sumThatSupportInfiniteInputs(10, 2.5, 3.14, 2 + 2j))
```

```python
import math
from typing import Literal, Tuple, Union
def calculate3dPointsDistance(point1: Tuple[Union[float, int], Union[float, int], Union[float, int]], point2: Tuple[Union[float, int], Union[float, int], Union[float, int]], Coordinate: Literal['Euclidean', 'Spherical']) -> float:
    """_summary_

    Parameters
    ----------
    point1 : Tuple[Union[float, int], Union[float, int], Union[float, int]]
        The coordinates of the first point.
    point2 : Tuple[Union[float, int], Union[float, int], Union[float, int]]
        The coordinates of the second point.
    Coordinate : Literal[&#39;Euclidean&#39;, &#39;Spherical&#39;]
        coordinate system: Euclidean or Spherical

    Returns
    -------
    float
        the distance between the two input points
    """
    if Coordinate != "Euclidean" and Coordinate != "Spherical":
        raise TypeError("the coordinate system must be either  Euclidian or spherical")
    
    if not isinstance(point1, tuple):
        raise TypeError("Point1 should be a tuple")
    
    if len(point1) != 3:
        raise ValueError("Point1 should contain three element")
    
    for i in point1:
        if not isinstance(i, (float, int)):
            raise TypeError("Element inside point1 should be floats or ints")
        
    if not isinstance(point2, tuple):
        raise TypeError("Point2 should be a tuple")
    
    if len(point2) != 3:
        raise ValueError("Point2 should contain three element")
    
    for i in point2:
        if not isinstance(i, (float, int)):
            raise TypeError("Element inside point2 should be floats or ints")
        
    if Coordinate == "Euclidean":
        return math.dist(point1, point2)
    elif Coordinate == "Spherical":
        r1, theta1, phi1 = point1
        r2, theta2, phi2 = point2
        if r1 < 0 or r2 < 0:
            raise ValueError("radius should be greater than zero")
        EuclideanCoor1 = (r1 * math.sin(theta1) * math.cos(phi1), r1 * math.sin(theta1) * math.sin(phi1), r1 * math.cos(theta1))
        EuclideanCoor2 = (r2 * math.sin(theta2) * math.cos(phi2), r2 * math.sin(theta2) * math.sin(phi2), r1 * math.cos(theta2))
        return math.dist(EuclideanCoor1, EuclideanCoor2)

print(calculate3dPointsDistance((0, 0, 0), (3, 4, 0), 'Euclidean')) #should be 5
print(calculate3dPointsDistance((10, math.pi / 2, 0), (10, math.pi / 2, math.pi / 2), 'Spherical')) #should be around 14.14
```

```python
from typing import Dict
def personDictionary(firstName: str, lastName: str, **info) -> Dict:
    """creating a dictionary for a person according to the name and information passed to this function as arguments

    Parameters
    ----------
    firstName : str
        input first name
    lastName : str
        input last name

    Returns
    -------
    Dict
        return dictionary containing this person's info
    """
    if not isinstance(firstName, str) or not isinstance(lastName, str):
        raise TypeError("first name and last name should be string")
    typesExpected = {'age': (int, float), 'occupation': str, 'location': str}
    for key, value in info.items():
        if key in typesExpected:
            if not isinstance(value, typesExpected[key]):
                raise TypeError(f"the {value} does not match the type of {key}, its type should be {typesExpected[key]}")
    
    result = {"first name": firstName, "last name": lastName}

    for key, value in info.items():
        result[key] = value

    return result

print(personDictionary("Tony", "He", age=18, occupation="student", location="UIUC"))
try:
    print(personDictionary("Tony", "He", age=18, occupation=2, location="UIUC"))
except TypeError:
    print("the error of ocupation is 2 is catched")
```
