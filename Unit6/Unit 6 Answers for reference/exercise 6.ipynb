```python
import matplotlib.pyplot as plt
import csv

with open("../Tutorial6.csv", "r") as f:
   next(csv.reader(f))
   next(csv.reader(f))
   #this is used to skip the first two line in the csv file

   wavenumbers = []
   transmittance = []


   for row in csv.reader(f):
      x = float(row[0])
      y = float(row[1])
      wavenumbers.append(x)
      transmittance.append(y)

   plt.figure(figsize=(10, 6))
   plt.plot(wavenumbers, transmittance, color='blue', linewidth=1)

   plt.xlabel(r'Wavenumber ($cm^{-1}$)')
   plt.ylabel('Transmittance (%T)')

   plt.gca().invert_xaxis()
```

```python
import matplotlib.pyplot as plt
import pandas as pd


df = pd.read_csv("../Tutorial6.csv", header=1)
#header=1 used to start from line 2 to skip the first line


plt.figure(figsize=(10, 6))
plt.plot(df['cm-1'], df['%T'], color='blue', linewidth=1)

plt.xlabel(r'Wavenumber ($cm^{-1}$)')
plt.ylabel('Transmittance (%T)')

plt.gca().invert_xaxis()

```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv("../SADExample.csv")



plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df, 
    x="ee (%)", 
    y="Alkene Type", 
    hue="Alkene Type",
    hue_order=['mono', 'gem_di', 'trans_di', 'cis_di', 'tri', 'tetra'],
    #i add hue because the error message "Passing `palette` without assigning `hue` is deprecated and will be removed in v0.14.0. Assign the `y` variable to `hue` and set `legend=False` for the same effect."
    order=['mono', 'gem_di', 'trans_di', 'cis_di', 'tri', 'tetra'],
    palette=['blue', 'orange', 'green', 'red', 'purple', 'brown'],
    inner="box",
    cut=0
    #cut is used to not allow the plot to stretch over its range
)

plt.xlim(-15, 115)
plt.title("Asymmetric Dihydroxylation database 1000 Entries")
plt.xlabel("ee (%)")
plt.ylabel("Alkene Type")
```

```python
import numpy as np


def NumpySortBySecondLetter(names):
    
    arr = np.array(names)

    get_second_letter = np.vectorize(lambda x: x[1])

    second_letters = get_second_letter(arr)

    sort_indices = np.argsort(second_letters, kind='stable')

    return arr[sort_indices]


print(NumpySortBySecondLetter(['Matthew', 'Tatsuto', 'Dongchen', 'Tony', 'Blake', 'Elena', 'Shangheng', 'Margherita', 'Matt', 'Jim', 'Nick', 'Jacob', 'Roberto', 'Felix', 'Michela', 'Yuki', 'Behrad', 'Sateesh'])) 
```
