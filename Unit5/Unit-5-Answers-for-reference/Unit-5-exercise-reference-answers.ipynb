```python
import math
from typing import Union
#exercise 1
class Point3D:
    def __init__(self, x: Union[int, float], y: Union[int, float], z: Union[int, float]) -> None:
        for arg in (x, y, z):
            if not isinstance(arg, (int, float)):
                raise TypeError(f"{arg} should be a numeric value")

        self.x = x
        self.y = y
        self.z = z
    def __sub__(self, other) -> float:
        if not isinstance(other, Point3D):
            raise TypeError("Can only subtract another Point3D object.")
        return math.dist((self.x, self.y, self.z), (other.x, other.y, other.z))
    
    def __eq__(self, other) -> bool:
        if not isinstance(other, Point3D):
            return False
        return (self - other) < 1e-6
    
    def __repr__(self) -> str:
        return f"Point3D(x={self.x:.3f}, y={self.y:.3f}, z={self.z:.3f})"
    
    #exercise 3
    @classmethod
    def create_from_string(cls, x: str, y: str, z: str):
        print("a typical use of a class method is to overload constructor")
        return cls(float(x), float(y), float(z))


    @classmethod
    def show_difference_class(cls) -> None:
        print("this is a class method")

    
    def show_difference_instance(self) -> None:
        print("this is a instance method")
    
#exercise 4

class Point3DWithMass(Point3D):
    def __init__(self, x: Union[int, float], y: Union[int, float], z: Union[int, float], mass: Union[int, float]) -> None:
        super().__init__(x,y,z)
        self.mass = mass

    @property
    def mass(self) -> Union[int, float]:
        return self._mass
    
    @mass.setter
    def mass(self, val) -> None:
        if not isinstance(val, (int, float)):
            raise TypeError("Mass must be a numeric value.")
        if val < 0:
            raise TypeError("imput value should not be negative")
        
        self._mass = val

    def __repr__(self) -> str:
        return f"Point3DWithMass(x={self.x:.3f}, y={self.y:.3f}, z={self.z:.3f}, mass={self.mass:.3f})"


#test for exercise 3
Point3D.show_difference_class()
instancePoint = Point3D(1, 2, 3)
instancePoint.show_difference_instance()
instancePointFromStr = Point3D.create_from_string('1', '2', '3')


```

```python
import math
from dataclasses import dataclass

#exercise 2

@dataclass
class Point3D:
    x: Union[float, int]
    y: Union[float, int]
    z: Union[float, int]
    def __post_init__(self):
        for arg in (self.x, self.y, self.z):
            if not isinstance(arg, (int, float)):
                raise TypeError(f"{arg} should be a numeric value")

    def __sub__(self, other) -> float:
        if not isinstance(other, Point3D):
            raise TypeError("Can only subtract another Point3D object.")
        return math.dist((self.x, self.y, self.z), (other.x, other.y, other.z))
    def __eq__(self, other) -> bool:
        if not isinstance(other, Point3D):
            return False
        return (self - other) < 1e-6
    def __repr__(self) -> str:
        return f"Point3D(x={self.x:.3f}, y={self.y:.3f}, z={self.z:.3f})"
    
```
